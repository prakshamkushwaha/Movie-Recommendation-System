1.Explain what Word2Vec is: It is a predictive machine learning model that uses a shallow, two-layer neural network to learn word associations and generate word embeddings from a large text corpus.

2.Predicting words from context: This relies on the distributional hypothesis: words appearing in similar contexts share similar meanings. By training the network to guess a missing word based on its neighbors (or vice versa), the internal weights adjust to capture the word's true meaning.

3.Define:
  Vocabulary: The total set of unique words present across the entire training dataset. 
  Context window: The specific number of words to look at on the left and right of the target word during training.   
  Embedding dimension: The fixed length of the dense vector representing each word (commonly 100, 200, or 300).
