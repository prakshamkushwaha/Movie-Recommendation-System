```python
import pandas as pd
import time
from nltk.tokenize import word_tokenize
from gensim.models import Word2Vec

# Load and prepare data (condensed)
df = pd.read_csv("IMDB Dataset.csv")
sentences = [[w for w in word_tokenize(r.lower()) if w.isalnum()] for r in df['review'].head(1000).astype(str)]

# 1. Train CBOW (for comparison)
start_cbow = time.time()
model_cbow = Word2Vec(sentences, vector_size=100, window=5, min_count=1, sg=0)
cbow_time = time.time() - start_cbow

# Train Skip-Gram (sg=1)
start_sg = time.time()
model_sg = Word2Vec(sentences, vector_size=100, window=5, min_count=1, sg=1)
sg_time = time.time() - start_sg

# 2. Compare Training time and Similar words output
print(f"Training Time - CBOW: {cbow_time:.2f}s | Skip-Gram: {sg_time:.2f}s")
print("\nCBOW Similar to 'good':", [w for w, s in model_cbow.wv.most_similar('good', topn=3)])
print("Skip-Gram Similar to 'good':", [w for w, s in model_sg.wv.most_similar('good', topn=3)])
```
