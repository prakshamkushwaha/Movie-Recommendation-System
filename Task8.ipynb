```python
import pandas as pd
from nltk.tokenize import word_tokenize
from gensim.models import Word2Vec

# Load and train model quickly
df = pd.read_csv("IMDB Dataset.csv")
sentences = [[w for w in word_tokenize(r.lower()) if w.isalnum()] for r in df['review'].head(2000).astype(str)]
model = Word2Vec(sentences, vector_size=100, window=5, min_count=5, sg=1)

# 1. Find most similar words to a given word
print("Words most similar to 'film':")
print(model.wv.most_similar('film', topn=3))

# 2 & 3. Perform vector arithmetic: king - man + woman = queen
# Note: IMDB might not have enough standard uses of king/queen. 
try:
    result = model.wv.most_similar(positive=['king', 'woman'], negative=['man'], topn=1)
    print(f"\nking - man + woman = {result[0][0]}")
except KeyError:
    print("\n'king', 'man', or 'woman' not in vocabulary subset.")

# 4. Test similar relationships from your dataset (e.g., actor - man + woman = actress)
try:
    res2 = model.wv.most_similar(positive=['actor', 'woman'], negative=['man'], topn=1)
    print(f"actor - man + woman = {res2[0][0]}")
except KeyError:
    print("Words missing for actor relationship test.")
```
