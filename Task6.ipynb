```python
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize
from gensim.models import Word2Vec

# Load and prepare data (condensed)
df = pd.read_csv("IMDB Dataset.csv")
sentences = [[w for w in word_tokenize(r.lower()) if w.isalnum()] for r in df['review'].head(1000).astype(str)]

# 1. Train a Word2Vec model with specified parameters (CBOW: sg=0)
model_cbow = Word2Vec(
    sentences=sentences, 
    vector_size=100, 
    window=5, 
    min_count=1, 
    sg=0
)

# 2. Print Vocabulary size and Embedding vector for a sample word
print(f"Vocabulary size (CBOW): {len(model_cbow.wv.key_to_index)}")
print(f"Embedding vector shape for 'movie': {model_cbow.wv['movie'].shape}")
print(f"Sample vector values for 'movie':\n{model_cbow.wv['movie'][:5]}")
```
