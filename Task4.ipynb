Input layer representation: A one-hot encoded vector representing the target word (Skip-Gram) or the average of one-hot vectors of the context words (CBOW).

Hidden layer (embedding layer): A linear layer without an activation function. The number of neurons here equals the requested embedding dimension (e.g., 100).

Output layer: A softmax classifier layer that outputs the probability distribution over the entire vocabulary, guessing the correct context or target word.

How weights become word embeddings: The weight matrix connecting the input layer to the hidden layer acts as a lookup table. After training, the row corresponding to a word's index in this matrix is directly extracted and used as that word's embedding vector.
