What are word embeddings? Word embeddings are dense, continuous-valued vector representations of words. They capture semantic and syntactic meanings by placing words with similar meanings closer together in the vector space.

Why one-hot encoding and BoW fail to capture semantics? They treat every word as an independent, isolated entity with orthogonal vectors (dot product of zero), meaning the model cannot recognize that "good" and "great" are related. They also result in highly sparse and high-dimensional matrices.

How word embeddings solve these problems? By projecting words into a dense, lower-dimensional space (e.g., 100 dimensions) where the vector's continuous values represent learned semantic features based on actual word co-occurrence in the text.
