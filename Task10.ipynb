Difference between CBOW & Skip-Gram in practice: CBOW averages context words and trains faster, making it suitable for massive datasets. Skip-Gram pairs the target with each context word individually, training slower but capturing better representations for rare and infrequent words.

Advantages of Word2Vec over TF-IDF: Word2Vec captures actual semantic meaning (synonyms have close vectors) and generates dense, fixed-size vectors (e.g., 100 dimensions) regardless of vocabulary size, whereas TF-IDF relies entirely on exact word matches and generates massive, sparse vectors.

Limitations of Word2Vec: It cannot handle Out-Of-Vocabulary (OOV) words. It also generates only one static vector per word, meaning it cannot differentiate between polysemous words (e.g., "bank" as a river vs. "bank" as a financial institution).

Why context still matters in modern NLP (lead-in to transformers): Because words change meaning depending on how they are used in a sentence. Modern models (like BERT) create contextual embeddings, where the vector for "bank" dynamically changes based on the entire sentence surrounding it, solving Word2Vec's static limitation.
