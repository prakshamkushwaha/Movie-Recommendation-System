```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from nltk.tokenize import word_tokenize
from gensim.models import Word2Vec

# Train model
df = pd.read_csv("IMDB Dataset.csv")
sentences = [[w for w in word_tokenize(r.lower()) if w.isalnum()] for r in df['review'].head(1000).astype(str)]
model = Word2Vec(sentences, vector_size=100, window=5, min_count=1, sg=1)

target_words = ['movie', 'film', 'actor', 'actress', 'good', 'great', 'bad', 'terrible']
valid_words = [w for w in target_words if w in model.wv]

# 1. Reduce embedding dimensions using PCA
word_vectors = np.array([model.wv[w] for w in valid_words])
pca = PCA(n_components=2)
twodim_vectors = pca.fit_transform(word_vectors)

# 2 & 3. Plot word embeddings in 2D and observe clusters
plt.figure(figsize=(8, 6))
plt.scatter(twodim_vectors[:, 0], twodim_vectors[:, 1], c='red')

for i, word in enumerate(valid_words):
    plt.annotate(word, xy=(twodim_vectors[i, 0] + 0.05, twodim_vectors[i, 1] + 0.05))

plt.title("PCA Visualization of Word2Vec Embeddings")
plt.grid(True)
plt.show() # In a Jupyter notebook, this will display the graph inline.
```
