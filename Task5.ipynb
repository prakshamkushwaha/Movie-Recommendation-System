```python
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize

# Run once if needed: nltk.download('punkt')

df = pd.read_csv("IMDB Dataset.csv")
# Using a subset of 1000 reviews for faster processing
text_data = df['review'].head(1000).astype(str).tolist()

# 1, 2 & 3. Tokenize cleaned text into sentences, then words, store as list of lists
sentences = []
for review in text_data:
    words = word_tokenize(review.lower())
    # Keep only alphanumeric tokens
    words = [w for w in words if w.isalnum()]
    if words:
        sentences.append(words)

# Example output
print(f"Total sentences prepared: {len(sentences)}")
print(f"Example (first sequence): {sentences[0][:6]}")
```
