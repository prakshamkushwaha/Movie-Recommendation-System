**1. CBOW (Continuous Bag of Words)**
* **Explanation:** Predicts a single target word from its surrounding context words.
* **Example:** The model takes the surrounding context words `["really", "love", "good", "movies"]` as its input and is trained to predict the target word `"watching"`.
* **Characteristics:** Faster to train and good for large datasets because it averages the context words together into a single observation.

**2. Skip-Gram**
* **Explanation:** Predicts the surrounding context words given a single target word.
* **Example:** The model takes the target word `"watching"` as its input and is trained to predict the surrounding context words `["really", "love", "good", "movies"]`. It does this by creating individual training pairs like `("watching", "love")` and `("watching", "good")`.
* **Characteristics:** Better for rare words and small datasets because it treats each context-target pair as a distinct new observation.
